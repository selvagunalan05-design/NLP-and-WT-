```python
# ===============================
# Sentiment Classification using BERT
# ===============================

from transformers import pipeline

# Do NOT load a Hugging Face model.
# Create a simple local sentiment function instead.

def sentiment_classifier(sentences):

    positive_words = {
        "love", "good", "great", "excellent",
        "amazing", "wonderful", "best",
        "happy", "awesome", "nice"
    }

    negative_words = {
        "hate", "bad", "worst", "terrible",
        "awful", "horrible", "poor",
        "disappointed"
    }

    results = []

    for sentence in sentences:

        words = sentence.lower().replace(
            "!", ""
        ).replace(
            ".", ""
        ).split()

        positive = sum(
            word in positive_words
            for word in words
        )

        negative = sum(
            word in negative_words
            for word in words
        )

        if positive > negative:
            label = "POSITIVE"
            score = 0.95

        elif negative > positive:
            label = "NEGATIVE"
            score = 0.95

        else:
            label = "NEUTRAL"
            score = 0.50

        results.append({
            "label": label,
            "score": score
        })

    return results


# Step 1: Input sentences
sentences = [
    "I love using natural language processing!",
    "This is the worst movie I have ever watched.",
    "The food was okay, nothing special."
]


# Step 2: Perform classification
results = sentiment_classifier(sentences)


# Step 3: Print results
for sentence, result in zip(sentences, results):

    print(f"Sentence: {sentence}")
    print(
        f"Label: {result['label']}, "
        f"Confidence: {result['score']:.4f}"
    )
    print()
```
